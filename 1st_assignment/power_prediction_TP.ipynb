```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Workpiece parameters
L = 40            # mm
Ds = 46           # mm
De = 24           # mm

# Tool parameters
ap = 1            # mm
f = 0.2           # mm/rev
Vc_const = 150    # m/min
Kc_vc = 3025.6    # MPa
N_const = 1200    # rpm
Kc_n = 3230.3     # MPa
```

```python
n = int((46 - 24) / (2*ap))  # number of double passes
D = np.arange(Ds, De - ap, -2*ap)    # mm

# Power for vc constant
Fc_vc = Kc_vc * ap * f                   # N
Pc_vc = np.full(n, Fc_vc * Vc_const / 60)   # W

# Power for n constant
Vc = np.pi * D * N_const / 1000              # m/min
Fc_n = Kc_n * ap * f                          # N
Pc_n = Fc_n * Vc / 60                          # W
```
